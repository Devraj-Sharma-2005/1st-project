```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Load the reduced CSV dataset
csv_path = '/content/drive/MyDrive/Minor Project\\/mp_rice_reduced.csv'
df = pd.read_csv(csv_path)

# 2. Separate target variable (y) and predictor features (X)
X = df.drop(columns=['rice_yield_kgha'])
y = df['rice_yield_kgha']

# 3. Perform One-Hot Encoding on categorical 'district' column
X_encoded = pd.get_dummies(X, columns=['district'], drop_first=True)

# 4. Split dataset into 80% Training and 20% Testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42
)

# 5. Output dataset shapes
print("--- DATASET SPLIT COMPLETE ---")
print(f"Total Features (after encoding): {X_encoded.shape[1]}")
print(f"Training Set (80%): {X_train.shape[0]} rows, {X_train.shape[1]} columns")
print(f"Testing Set  (20%): {X_test.shape[0]} rows, {X_test.shape[1]} columns")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1. Initialize models
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
}

# 2. Train and evaluate each model
results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)

    results.append({
        "Model": name,
        "R2 Score": round(r2, 4),
        "RMSE (kg/ha)": round(rmse, 2),
        "MAE (kg/ha)": round(mae, 2)
    })

# 3. Display benchmarking results
results_df = pd.DataFrame(results)
print("--- MODEL BENCHMARK RESULTS ---")
display(results_df)

# 4. Plot Actual vs Predicted Yield for best model (Gradient Boosting)
best_model = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
best_model.fit(X_train, y_train)
y_pred_best = best_model.predict(X_test)

plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred_best, alpha=0.7, color='teal')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.title("Actual vs Predicted Rice Yield (kg/ha)", fontsize=12, fontweight='bold')
plt.xlabel("Actual Yield (kg/ha)")
plt.ylabel("Predicted Yield (kg/ha)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('/content/drive/MyDrive/Minor Project\\/actual_vs_predicted.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Extract feature importances from the best model (Gradient Boosting)
feature_importances = best_model.feature_importances_
feature_names = X_train.columns

# 2. Create a DataFrame for visual mapping
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': feature_importances
}).sort_values(by='Importance', ascending=False)

# 3. Plot top 10 features
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Importance',
    y='Feature',
    data=importance_df.head(10),
    palette='viridis'
)

plt.title("Top 10 Most Important Features for Rice Yield Prediction", fontsize=12, fontweight='bold')
plt.xlabel("Relative Importance Score")
plt.ylabel("Feature")
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('/content/drive/MyDrive/Minor Project\\/feature_importance.png', dpi=300, bbox_inches='tight')
plt.show()

print("--- TOP 5 DRIVING FACTORS ---")
display(importance_df.head(5))
```

```python
import joblib
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# 1. Load trained model and reduced dataset structure
model = joblib.load('/content/drive/MyDrive/Minor Project\\/rice_yield_model.pkl')
csv_path = '/content/drive/MyDrive/Minor Project\\/mp_rice_reduced.csv'
df = pd.read_csv(csv_path)

# Extract unique districts
districts = sorted(df['district'].unique().tolist())

# 2. Build Interactive UI Controls
district_dropdown = widgets.Dropdown(options=districts, description='District:', value='jabalpur')
area_slider = widgets.FloatSlider(value=50.0, min=0.0, max=300.0, step=1.0, description='Rice Area (k ha):')
canal_slider = widgets.FloatSlider(value=20.0, min=0.0, max=200.0, step=1.0, description='Canal Irrig (k ha):')
tubewell_slider = widgets.FloatSlider(value=30.0, min=0.0, max=200.0, step=1.0, description='Tubewell Irrig:')
nitrogen_slider = widgets.FloatSlider(value=80.0, min=0.0, max=200.0, step=1.0, description='Nitrogen (kg/ha):')
tmin_slider = widgets.FloatSlider(value=24.0, min=15.0, max=35.0, step=0.5, description='Monsoon Tmin (°C):')

predict_button = widgets.Button(description="Predict Rice Yield", button_style='success')
output = widgets.Output()

# 3. Define Prediction Function
def predict_yield(b):
    with output:
        output.clear_output()

        # Prepare template row matching One-Hot Encoded features
        X_template = pd.get_dummies(df.drop(columns=['rice_yield_kgha']), columns=['district'], drop_first=True)
        input_data = pd.DataFrame(0, index=[0], columns=X_template.columns)

        # Set numerical inputs
        input_data['year'] = 2026
        input_data['rice_area_1000ha'] = area_slider.value
        input_data['irrig_canal_area_1000ha'] = canal_slider.value
        input_data['irrig_tubewell_area_1000ha'] = tubewell_slider.value
        input_data['irrig_tank_area_1000ha'] = 5.0
        input_data['n_per_ha_nca'] = nitrogen_slider.value
        input_data['p_per_ha_nca'] = 30.0
        input_data['k_per_ha_nca'] = 15.0
        input_data['rain_monsoon_mm'] = 900.0
        input_data['rain_annual_mm'] = 1100.0
        input_data['tmax_monsoon'] = 32.0
        input_data['tmin_monsoon'] = tmin_slider.value
        input_data['lgp_days'] = 120
        input_data['water_stress_score'] = 0.2
        input_data['soil_vertisols'] = 1
        input_data['soil_inceptisols'] = 0

        # Set categorical district feature
        selected_dist = district_dropdown.value
        dist_col = f"district_{selected_dist}"
        if dist_col in input_data.columns:
            input_data[dist_col] = 1

        # Predict yield
        predicted_yield = model.predict(input_data)[0]

        print(f"🎯 Estimated Rice Yield for {selected_dist.title()}: {predicted_yield:.2f} kg/ha")

predict_button.on_click(predict_yield)

# Display Form
display(district_dropdown, area_slider, canal_slider, tubewell_slider, nitrogen_slider, tmin_slider, predict_button, output)
```
